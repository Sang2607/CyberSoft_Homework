# **VI. Date Type in Pandas**

```python
dateR = pd.period_range('01/01/2026',freq='30d',periods=6)
dateR
```

```python
#Lấy ngày đầu tiên
dateR.min()
```

```python
#Lấy ngày gần nhất
dateR.max()
```

```python
#Biến thành Dataqframe:
date_df = pd.DataFrame(data=dateR,columns=['sample_date'])
date_df
```

```python
#Lấy ngày trước và sau của ô hiện tại
date_df['prev_date'] = date_df['sample_date'].shift(periods=1)
date_df
```

```python
date_df['next_date'] = date_df['sample_date'].shift(periods=-1)
date_df
```

```python
#Cách 1
date_df['diff_date'] = date_df['next_date'] - date_df['sample_date']
#Cách 2
date_df["date_diff"] = date_df["sample_date"].diff(periods=1)

date_df

```

```python
#Tìm ngày đầu tiên trong tháng
date_df['fist_day_month'] = date_df['sample_date'].values.astype("datetime64[M]")
date_df
```

```python
#Tìm thứ trong tuần:
date_df = pd.DataFrame(data=dateR,columns=['sample_date'])
date_df['sample_date'] = date_df['sample_date'].dt.to_timestamp()
date_df['sample_date'].dt.day_name()
```

# ** VIII.Map and ApplyMap trong data **

```python
df = pd.DataFrame({"Region":['North','West','East','South','North','West','East','South'],
          "Team":['One','One','One','One','Two','Two','Two','Two'],
          "Squad":['A','B','C','D','E','F','G','H'],
          "Revenue":[7500,5500,2750,6400,2300,3750,1900,575],
            "Cost":[5200,5100,4400,5300,1250,1300,2100,50]})
#Tạo ra 1 cột Profit: Revenue > Cost : profit | loss
import numpy as np
#cách 1:
#df['Profit'] = np.where(df['Revenue'] > df['Cost'],"Profit","Loss")

#cách 2:
df['Profit'] = df.apply(lambda x: 'Profit' if x['Revenue'] > x['Cost'] else 'Loss',axis=1) #axis = 1 là xét trên cột
df
```

```python
df['Color'] = np.where(df['Team'] == 'One',"Red","Blue")
df
```

```python
team_map =  {"One":"Red","Two":"Blue"}
df['Team_Color'] = df['Team'].map(team_map)
df
```

```python
df.apply(lambda x: len(x))
```

```python
pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
df ['Region_total'] = df.groupby("Region")['Revenue'].transform('sum')
df
```
