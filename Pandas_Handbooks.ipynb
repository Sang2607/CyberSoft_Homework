## **VI. Date Type in Pandas**

```python
daterange = pd.period_range('1/1/2026', freq='30d', periods=5)
daterange
```

```python
#Lấy ngày đầu tiên trong chuỗi: #The first date 
daterange.min()
```

```python
#Lấy ngày gần nhất: #The most current date
daterange.max()
```

```python
#Biến thành DataFrame:
date_df = pd.DataFrame(data=daterange, columns=['sample_date'])
date_df
```

```python
#Lấy ngày trước và sau của ô hiện tại
#Ordered_Date -> ShippedDate
##Logistics: Leadtime: ShippedDate - Order_Date = ?
date_df = pd.DataFrame(data=daterange, columns=['sample_date'])
date_df['prev_date'] = date_df['sample_date'].shift(periods=1)
date_df['next_date'] = date_df['sample_date'].shift(periods=-1)
#date_df['diff date'] = date_df['next_date'] - date_df['sample_date']
date_df['date diff'] = date_df['sample_date'].diff(periods=1)
date_df
```

```python
#Tìm ngày đầu tiên trong tháng:
date_df['first of month'] = date_df['sample_date'].values.astype('datetime64[M]')
date_df

#GROUP BY
#2026-07-18
#2026-07-13
#-> 2026-07-01: Data của kỳ xét tháng 07/2026
```

## **VIII. Map and ApplyMap trong Pandas**

```python
df = pd.DataFrame({"Region": ['North','West','East','South','North','West','East','South'],
          "Team": ['One','One','One','One','Two','Two','Two','Two'],
          "Squad": ['A','B','C','D','E','F','G','H'],
          "Revenue":[7500,5500,2750,6400,2300,3750,1900,575],
            "Cost":[5200,5100,4400,5300,1250,1300,2100,50]})
df
```

```python
import numpy as np
```

```python
#Tạo 1 cột Profit: Revenue > Cost : Profit | Loss 
df['Profit'] = np.where(df['Revenue'] > df['Cost'], "Profit", "Loss")
df
```

```python
df['Profit'] = df.apply(lambda x: 'Profit' if x['Revenue'] > x['Cost'] else 'Loss', axis =1) #axis =1 là xét trên cột
```

```python
# Tạo cột Color: Team One -> RED, Team Two -> Blue
df['Color'] = np.where(df['Team'] == 'One', "Red", "Blue")
df
```

```python
team_map = {"One": "Red", "Two": "Blue"}
df['Team_Color'] = df['Team'].map(team_map)
df
```

```python
df.apply(lambda x: len(str(x)))
```

```python
df
```

```python
#Group BY 
df['Region_Revenue_Total'] = df.groupby('Region')['Revenue'].transform('sum')
df
```

```python
df['Contribution_Percentage'] = df['Revenue'] / df['Region_Revenue_Total']
df
```

## **X. Merge DataFrame**

```python
df1 = pd.DataFrame({'letter': ['A', 'B', 'C', 'D'],
                    'number': [1, 2, 3, 4]})
df2 = pd.DataFrame({'letter': ['C', 'D', 'E', 'F'],
                    'number': [3, 4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
df1.merge(df2, how='left', on ='number')
```

```python
df1.merge(df2, how = 'inner', on = 'number')
```

```python
df1.merge(df2, how='right', on ='number')
```

```python
pd.concat([df1, df2], axis =1) #chiều ngang
```

```python
pd.concat([df1, df2], axis =0).drop_duplicates() #chiều dọc
```
